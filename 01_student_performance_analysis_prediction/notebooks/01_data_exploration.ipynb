# 01 — Data Exploration

This notebook explores the UCI Student Performance dataset. Run `python -m src.download_data` first.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd()
DATA_PATH = ROOT / 'data' / 'raw' / 'student_performance.csv'
if not DATA_PATH.exists():
    raise FileNotFoundError('Run python -m src.download_data from the repository root first.')

df = pd.read_csv(DATA_PATH)
print('Shape:', df.shape)
df.head()
```

## 1. Data structure

Inspect column types, non-null values and summary statistics. Do not assume every numeric-coded column is continuous; some are ordinal categories.

```python
df.info()
df.describe(include='all').T
```

## 2. Data quality

```python
print('Missing values by column:')
display(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## 3. Target distribution

`G3` is the final grade. Describe its distribution before fitting models.

```python
display(df['G3'].describe())
plt.figure(figsize=(9, 5))
sns.histplot(df['G3'], bins=20, kde=True)
plt.title('Distribution of Final Grade (G3)')
plt.xlabel('Final grade')
plt.ylabel('Number of records')
plt.tight_layout()
plt.show()
```

## 4. Numeric correlations

Correlation measures linear association, not causation. Also note that G1 and G2 are prior grades and should be excluded from an early-prediction model.

```python
numeric = df.select_dtypes(include='number')
plt.figure(figsize=(12, 9))
sns.heatmap(numeric.corr(), cmap='coolwarm', center=0)
plt.title('Correlation Heatmap — Numeric Columns')
plt.tight_layout()
plt.show()
```
